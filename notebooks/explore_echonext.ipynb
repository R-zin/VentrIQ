```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("NumPy:", np.__version__)
print("Pandas:", pd.__version__)
```

```python
DATA_DIR = Path("../data/raw/echonext")

print(DATA_DIR.resolve())
```

```python
files = [
    "echonext_metadata_100k.csv",
    "EchoNext_train_waveforms.npy",
    "EchoNext_val_waveforms.npy",
    "EchoNext_test_waveforms.npy",
    "EchoNext_train_tabular_features.npy",
    "EchoNext_val_tabular_features.npy",
    "EchoNext_test_tabular_features.npy",
]

for name in files:
    path = DATA_DIR / name

    if path.exists():
        size_gb = path.stat().st_size / (1024 ** 3)
        print(f"✓ {name:45} {size_gb:.2f} GB")
    else:
        print(f"✗ {name}")
```

```python
metadata = pd.read_csv(
    DATA_DIR / "echonext_metadata_100k.csv"
)

print("Shape:", metadata.shape)

metadata.head()
```

```python
print("Columns:")
print()

for i, column in enumerate(metadata.columns):
    print(f"{i:3}: {column}")
```

```python
print(metadata["lvef_value"].describe())
```

```python
plt.figure(figsize=(10, 5))

plt.hist(
    metadata["lvef_value"].dropna(),
    bins=50
)

plt.xlabel("LVEF (%)")
plt.ylabel("Number of ECGs")
plt.title("EchoNext LVEF Distribution")

plt.show()
```
